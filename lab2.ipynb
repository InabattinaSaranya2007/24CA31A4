1.Define tokenization with an example.

Tokenization:
Tokenization is the process of breaking a sentence or text into smaller units called tokens, such as words or sentences. It is the first step in text preprocessing for NLP.

Example:
Sentence: "Python is easy to learn."
Tokens: ['Python', 'is', 'easy', 'to', 'learn']

2.What are stop words? Give five examples.

Stop words are common words that appear frequently in text but usually do not add significant meaning. They are often removed during text preprocessing to improve analysis.

Examples (any five):

and

3.Define stemming. Give one example.

Stemming is the process of reducing a word to its root or base form by removing prefixes or suffixes. It helps group similar words together during text processing.

Example:

Playing → Play

4.What is lemmatization? How does it differ from stemming?

Lemmatization is the process of converting a word into its meaningful base form (called a lemma) using vocabulary and grammar rules.

Difference between Lemmatization and Stemming:

Stemming: Removes prefixes or suffixes to get the root word. The result may not be a valid English word.
Lemmatization: Converts a word to its actual dictionary (base) form. The result is always a meaningful word.

Example:

Stemming: studies → studi
Lemmatization: studies → study

5.What is Part-of-Speech (POS) tagging?

**Part-of-Speech (POS) Tagging:**
Part-of-Speech (POS) tagging is the process of assigning a grammatical category to each word in a sentence, such as **noun, verb, adjective,** or **adverb**. It helps the computer understand the role of each word in the sentence.

**Example:**
Sentence: **"The dog barks."**

* The → Determiner (DT)
* dog → Noun (NN)
* barks → Verb (VB)

```python
# LAB 2 – STEP 5
# PART-OF-SPEECH (POS) TAGGING
# ============================================================

# Import Libraries
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download Resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

# For NLTK 3.9+
# nltk.download('punkt_tab')
# nltk.download('averaged_perceptron_tagger_eng')

# Input Sentence
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."

# Step 1 : Tokenization
tokens = word_tokenize(text)

# Step 2 : Stop-word & Punctuation Removal
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]

# Step 3 : Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = []

for word in filtered_tokens:

    if word.lower() == "studying":
        lemma = lemmatizer.lemmatize(word.lower(), pos="v")
    else:
        lemma = word

    lemmatized_tokens.append(lemma)

# Step 4 : POS Tagging
pos_tags = nltk.pos_tag(lemmatized_tokens)

# Display Results
print("=" * 60)
print("Lemmatized Tokens")
print("=" * 60)
print(lemmatized_tokens)

print("\n" + "=" * 60)
print("POS Tagged Tokens")
print("=" * 60)

for word, tag in pos_tags:
    print(f"{word:<12} --> {tag}")

print("\n" + "=" * 60)
print("POS Tagged List")
print("=" * 60)
print(pos_tags)
```
